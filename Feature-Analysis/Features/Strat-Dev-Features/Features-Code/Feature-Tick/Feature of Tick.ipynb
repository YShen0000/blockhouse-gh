# This feature identify which side initiate the trade
- 1. b = 1：buyer
- 2. b = -1:seller

```python
import numpy as np
import pandas as pd
import yfinance as yf
import datetime
import matplotlib.pyplot as plt
```

```python
# Feature
def TickRule(data):
    result = pd.DataFrame()
    result['Datetime'] = data['Datetime']
    result['Tick'] = 1
    result.loc[(data['Close'].diff()<0) & (data['Datetime'].dt.date==data['Datetime'].shift().dt.date),'Tick'] = -1
    for idx in data.index[1:]:
        if data['Close'].diff().iloc[idx] == 0 and data['Datetime'].iloc[idx].date() == data['Datetime'].iloc[idx - 1].date():
            result.iloc[idx, 1] = result.iloc[idx - 1, 1]
    return result
```

```python
WFC = yf.download('WFC', interval='1m', period='1d')
WFC.columns = WFC.columns.droplevel(1)
WFC.reset_index(inplace=True)
TickRule(WFC)
```
