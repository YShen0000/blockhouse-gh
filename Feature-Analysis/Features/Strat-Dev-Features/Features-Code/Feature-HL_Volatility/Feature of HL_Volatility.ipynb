# This feature provides two ways to estimate volatility based on high-low prices, which is thought to be a better way than based on close price

```python
import numpy as np
import pandas as pd
import datetime
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Feature 1
def HLVol1(data):
    
    window_size = 780 # 2 days
    result = pd.DataFrame()
    result['Datetime'] = data['Datetime']
    
    result['HL_Vol1'] = (np.log(data['High'])-np.log(data['Low']))**2
    result['HL_Vol1'] = result['HL_Vol1'].rolling(window=window_size).mean()
    result['HL_Vol1'] = (result['HL_Vol1']/(4*np.log(2)))**0.5
    
    return result
```

```python
# Feature 2
def HLVol2(data):
    
    window_size = 780 # 2 days
    result = pd.DataFrame()
    result['Datetime'] = data['Datetime']
    
    result['HL_Vol2'] = np.log(data['High'])-np.log(data['Low'])
    result['HL_Vol2'] = result['HL_Vol2'].rolling(window=window_size).mean()
    result['HL_Vol2'] = result['HL_Vol2']/(8/np.pi)**0.5
    
    return result
```
