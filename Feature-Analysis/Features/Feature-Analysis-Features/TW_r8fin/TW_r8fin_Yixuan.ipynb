# Mapping Trade and Market Data

The market data is recorded at 10-second intervals, meaning that within each 10-second window, the bid and ask prices are assumed to remain constant. To align the trading data with the corresponding market conditions, we round the executiontime in the trading data to the 10 seconds. This ensures that each trade is matched with the correct bid and ask price available at that time interval.

```python
# Convert execution time to datetime format
trading_data['executiontime'] = pd.to_datetime(trading_data['executiontime'])

# Round execution time down to the nearest 10 seconds
trading_data['executiontime_10s'] = trading_data['executiontime'].dt.floor('10S')
```

```python
trading_data[['executiontime', 'symbol', 'executiontime_10s']]
```

Filter a subdataframe with only symbols in ['30 Year', '5 Year', '2 Year', '10 Year', '3 Year', '7 Year', '20 Year']

```python
# Define the list of symbols that are standard maturities and can be mapped to market data
maturity_symbols = ['30 Year', '5 Year', '2 Year', '10 Year', '3 Year', '7 Year', '20 Year']

# Filter the trading data to include only rows with the selected maturity symbols
filtered_trading_data = trading_data[trading_data['symbol'].isin(maturity_symbols)]
```

```python
# Define the list of symbols that are standard maturities and can be mapped to market data
maturity_symbols = ['30 Year', '5 Year', '2 Year', '10 Year', '3 Year', '7 Year', '20 Year']
market_symbols = ['30yr', '5yr', '2yr', '10yr', '3yr', '7yr', '20yr']

# Create a mapping between trading symbols and market symbols
symbol_mapping = dict(zip(maturity_symbols, market_symbols))

# Create an empty list to store merged data for each symbol
merged_data_list = []

# Loop through each symbol and merge with corresponding market data
for maturity_symbol, market_symbol in symbol_mapping.items():
    # Filter trading data for the current symbol
    trading_data_filtered = filtered_trading_data[filtered_trading_data['symbol'] == maturity_symbol]
    market_data_filtered = market_data[market_symbol].copy()

    # Convert market data time to datetime format
    market_data_filtered['time'] = pd.to_datetime(market_data_filtered['time'])

    # Calculate mid-price as (bid_price + ask_price) / 2
    market_data_filtered['mid_price'] = (market_data_filtered['bid_price'] + market_data_filtered['ask_price']) / 2

    # Add a column for mid-price at t + 10s (p_{t+10s}) by shifting the mid_price column
    market_data_filtered['mid_price_t_plus_10s'] = market_data_filtered['mid_price'].shift(-1)
    print(market_data_filtered)

    # Merge trading data with market data based on exact match of time, including bid/ask size and price
    merged_data = pd.merge(trading_data_filtered, market_data_filtered,
                           left_on='executiontime_10s', right_on='time',
                           how='left')

    # Append merged data to the list
    merged_data_list.append(merged_data)
```

```python
# Concatenate all merged data
mapped_data = pd.concat(merged_data_list, ignore_index=True)
```

### Calculate Price Variation

```python
mapped_data[['symbol', 'executiontime_10s', 'price', 'quantity']].head(50)
```

```python
# Calculate price variation as the difference between mid-price at t + 10s and t
mapped_data['price_variation'] = mapped_data['mid_price_t_plus_10s'] - mapped_data['mid_price']
```

```python
# Calculate price variation volatility for each symbol using a rolling window approach
def calculate_volatility(group):
    group = group.sort_values('executiontime').set_index('executiontime')
    group['price_variation_volatility'] = group['price_variation'].rolling('10s', closed='both').std().fillna(0)
    return group.reset_index()

mapped_data = mapped_data.groupby('symbol', group_keys=False, as_index=False).apply(calculate_volatility)
```

```python
mapped_data
```

```python
# Calculate signed order flow: buy orders as positive, sell orders as negative
mapped_data['signed_order_flow'] = mapped_data['side'].map({'B': 1, 'S': -1}) * mapped_data['quantity']

# Calculate order flow imbalance (net market order flow) for each 10s time window
order_flow_imbalance = mapped_data.groupby('executiontime_10s')['signed_order_flow'].sum().reset_index()
order_flow_imbalance.columns = ['executiontime_10s', 'order_flow_imbalance']
```

```python
order_flow_imbalance
```

```python
# Merge order flow imbalance back into mapped_data
mapped_data = pd.merge(mapped_data, order_flow_imbalance, on='executiontime_10s', how='left')
```

```python
mapped_data
```

```python
# Calculate Signed Order Flow Volatility for each 10-second window
signed_order_flow_volatility = mapped_data.groupby('executiontime_10s')['signed_order_flow'].std().reset_index()
signed_order_flow_volatility.columns = ['executiontime_10s', 'signed_order_flow_volatility']

# Merge signed order flow volatility back to mapped_data
mapped_data = pd.merge(mapped_data, signed_order_flow_volatility, on='executiontime_10s', how='left')

# Fill NaN values with 0 for volatility
mapped_data['signed_order_flow_volatility'] = mapped_data['signed_order_flow_volatility'].fillna(0)
```
