```python
def get_expected_price(symbol, timestamp):
    # Initialize the client with your API key
    try:
        timestamp = pd.to_datetime(timestamp)
    except Exception as e:
        print(f"Error parsing timestamp: {e}")
        return None, None, None, None, None

    client = db.Historical("db-UVp9SCqXjLpm5qprthxS335uppKae")

    # Convert timestamp to nanoseconds
    end_timestamp = int(timestamp.timestamp() * 1_000_000_000)
    start_timestamp = end_timestamp - (60 * 1_000_000_000)  # Subtract 1 minute from the timestamp for start time

    try:
        # Fetching Market by Price (MBP) data
        data = client.timeseries.get_range(
            dataset="XNAS.ITCH",  # Correct dataset for NASDAQ equities
            schema="mbp-1",       # Schema for Market by Price, top level only
            symbols=[symbol],
            start=start_timestamp,
            end=end_timestamp,
        )

        # Convert the data to a pandas DataFrame for processing
        df_mbp = data.to_df()

        if df_mbp.empty:
            print(f"No data found for the given timestamp: {timestamp}")
            return None, None, None, None, None

        # Drop rows with NaN values in the relevant columns
        df_mbp_cleaned = df_mbp.dropna(subset=['bid_px_00', 'ask_px_00', 'bid_sz_00', 'ask_sz_00'])

        if df_mbp_cleaned.empty:
            print(f"All data rows contained NaN values and were dropped for the given timestamp: {timestamp}")
            return None, None, None, None, None

        # Initialize variables to store maximum bid price and minimum ask price, and their sizes
        max_bid_price = float('-inf')
        min_ask_price = float('inf')
        bid_sizes = []
        ask_sizes = []
        bid_prices = []
        ask_prices = []

        # Process the cleaned data to find the maximum bid price and collect sizes
        for index, row in df_mbp_cleaned.iterrows():
            if row['bid_px_00'] > max_bid_price:
                max_bid_price = row['bid_px_00']
            if row['ask_px_00'] < min_ask_price:
                min_ask_price = row['ask_px_00']

            bid_prices.append(row['bid_px_00'])
            bid_sizes.append(row['bid_sz_00'] * 100)
            ask_prices.append(row['ask_px_00'])
            ask_sizes.append(row['ask_sz_00'] * 100)

        # Check if max_bid_price was updated, otherwise handle no data case
        if max_bid_price == float('-inf') or min_ask_price == float('inf'):
            print(f"No valid bid or ask prices found after cleaning the data for the given timestamp: {timestamp}")
            return None, None, None, None, None

        # Return five values
        return min_ask_price, bid_sizes, ask_sizes, bid_prices, ask_prices

    except Exception as e:
        print(f"An error occurred: {e}")
        return None, None, None, None, None


def get_twap_trades(env, target_shares, preferred_timeframe):
    data = env.data
    total_steps = len(data)
    twap_shares_per_step = target_shares / min(preferred_timeframe, total_steps)

    remaining_inventory = target_shares
    trades = []

    for step in range(min(total_steps, preferred_timeframe)):
        timestamp_str = data['Timestamp'].iloc[step]
        ticker_symbol = data['Ticker'].iloc[step]

        # Fetch the expected price, bid sizes, and ask sizes from the API
        expected_price, bid_sizes, ask_sizes, bid_prices, ask_prices = get_expected_price(ticker_symbol, timestamp_str)

        if expected_price is None:
            print(f"Skipping step {step} due to lack of data.")
            continue

        # Calculate the volume to trade for this step
        size_of_slice = min(twap_shares_per_step, remaining_inventory)
        remaining_inventory -= int(np.ceil(size_of_slice))

        # Use the VWAP price (based on ask prices) as the execution price for TWAP
        twap_price = calculate_vwap(ask_prices, ask_sizes)

        trade = {
            'step': step,
            'execution_price': twap_price,  # Using VWAP price from ask side for buy
            'shares': size_of_slice,
            'inventory': remaining_inventory,
            'action': (1, 1),  # 1 indicates 1 step taken, another 1 is unused but kept for compatibility
        }
        trades.append(trade)

    return pd.DataFrame(trades)


def get_vwap_trades(env, target_shares, preferred_timeframe):
    data = env.data
    total_volume = data['volume'].sum()
    total_steps = len(data)
    remaining_inventory = target_shares
    trades = []

    for step in range(min(total_steps, preferred_timeframe)):
        timestamp_str = data['Timestamp'].iloc[step]
        ticker_symbol = data['Ticker'].iloc[step]

        # Fetch the expected price, bid sizes, and ask prices from the API
        expected_price, bid_sizes, ask_sizes, bid_prices, ask_prices = get_expected_price(ticker_symbol, timestamp_str)

        if expected_price is None or not ask_prices or not ask_sizes:
            print(f"Skipping step {step} due to lack of data.")
            continue

        # Calculate the volume to trade for this step based on VWAP (ask side)
        volume_at_step = data['volume'].iloc[step]
        size_of_slice = (volume_at_step / total_volume) * target_shares
        size_of_slice = min(size_of_slice, remaining_inventory)
        remaining_inventory -= int(np.ceil(size_of_slice))

        # Calculate VWAP using ask prices and ask sizes
        vwap_price = calculate_vwap(ask_prices, ask_sizes)

        trade = {
            'step': step,
            'execution_price': vwap_price,  # VWAP calculated using ask prices for buy
            'shares': size_of_slice,
            'inventory': remaining_inventory,
            'action': (1, 1),  # 1 indicates 1 step taken, another 1 is unused but kept for compatibility
        }
        trades.append(trade)

    return pd.DataFrame(trades)


```

```python
def calculate_vwap(bid_prices, bid_sizes):
    """Calculate VWAP based on bid prices and bid sizes."""
    if not bid_prices or not bid_sizes or sum(bid_sizes) == 0:
        return 0.0
    weighted_sum = sum(price * size for price, size in zip(bid_prices, bid_sizes))
    total_volume = sum(bid_sizes)
    vwap = weighted_sum / total_volume
    return vwap
```

```python
# Load the trained model
model = PPO.load("./Models/trading_agent_v2.zip")

# Reset the environment to start inference
obs = env.reset()
total_slippage = 0.0  # Ensure these are floats
total_price_impact = 0.0  # Ensure these are floats
total_spread_cost = 0.0  # Ensure these are floats

trade_details = []  # Store trade details

for _ in range(len(env.data)):
    # Predict the action to take based on the current observation
    action, _states = model.predict(obs)
    
    # Step through the environment using the predicted action
    obs, rewards, done, info = env.step(action)
    
    # Get the current step and timestamp
    current_step = env.current_step
    timestamp_str = env.data['Timestamp'].iloc[current_step - 1]
    ticker_symbol = env.data['Ticker'].iloc[current_step - 1]

    # Fetch the expected price, bid sizes, and ask sizes from the API
    expected_price, bid_sizes, ask_sizes, bid_prices, ask_prices = env.get_expected_price(ticker_symbol, timestamp_str)
    
    if expected_price is None:
        print(f"Skipping step {current_step} due to lack of data.")
        continue

    # Calculate VWAP price
    vwap_price = calculate_vwap(bid_prices, bid_sizes)

    # Debugging prints
    #print(f"Expected Price: {expected_price}")
    #print(f"VWAP Price: {vwap_price}")

    # Extract the market price from the observation
    #market_price = obs[3] # Assuming 'close' price is at index 3
    market_price = vwap_price
    if market_price == 0.0:
        print(f"Skipping step {current_step} due to invalid market price.")
        continue
    
    # Extract the order type and limit price from the action
    order_type = "Market" if action[0] < 0.5 else "Limit"
    limit_price_scale = action[1]
    #limit_price_scale = limit_price_scale/5
    print(f'LIMIR SALEEEEEE:{limit_price_scale}')

    # Use VWAP price if the order type is a market order
    if order_type == "Market":
        execution_price = vwap_price
    else:
        limit_price = market_price + limit_price_scale * (0.05 * market_price)
        # implement resting order logic because if limit is more than best price then you cant sell
        #####################
        execution_price = limit_price

    # Extract the volume (shares) directly from the environment's data
    volume = env.data['shares'].iloc[current_step - 1]  # Using current_step - 1 to match the volume used in the step

    # Track metrics
    print(f"EXPECTEDD {expected_price}, executionnnn {execution_price}")
    slippage = abs(expected_price - execution_price)
    total_slippage += slippage * volume

    # Calculate price impact using the transaction cost from the data
    price_impact = env.data['transaction_cost'].iloc[current_step - 1] * volume
    total_price_impact += price_impact
    
    # Handle cancel actions
    cancel_actions = action[2:2 + env.max_orders]
    canceled_orders = [i for i, cancel in enumerate(cancel_actions) if cancel >= 0.5]
    
    # Calculate spread cost using the difference between ask and bid prices
    spread_cost = (np.min(ask_prices) - np.max(bid_prices)) * volume
    total_spread_cost += spread_cost

    # Append the trade details to the list
    trade_details.append({
        'step': current_step - 1,
        'Order Type': order_type,
        'Execution Price': execution_price,
        'Canceled Orders': canceled_orders,
        'Reward': rewards,
        'Volume': volume,
        'Slippage': total_slippage,
        'Price Impact': price_impact,
        'Spread Cost': spread_cost
    })
    
    print(f"Trade Executed: {order_type} Order at Price: {execution_price}, Volume: {volume}")
    if canceled_orders:
        print(f"Canceled Orders: {canceled_orders}")

    # If the environment is done, break the loop
    if done:
        
        break
#print(f'Trade Deets {trade_details}')
# Render the final state of the environment
env.render()

# At the end, before printing, let's add some debugging print statements
print(f"Total Slippage (Type): {type(total_slippage)}")
print(f"Total Price Impact (Type): {type(total_price_impact)}")
print(f"Total Spread Cost (Type): {type(total_spread_cost)}")

# Ensure the final output values are floats, not arrays
if isinstance(total_slippage, np.ndarray):
    total_slippage = total_slippage.item()  # Convert to scalar
if isinstance(total_price_impact, np.ndarray):
    total_price_impact = total_price_impact.item()  # Convert to scalar
if isinstance(total_spread_cost, np.ndarray):
    total_spread_cost = total_spread_cost.item()  # Convert to scalar

# Compute final metrics
print(f"Total Slippage: {total_slippage:.2f}")
print(f"Total Price Impact: {total_price_impact:.2f}")
print(f"Total Spread Cost: {total_spread_cost:.2f}")
```

```python
def compare_models(env, rl_trade_details, twap_trade_details, vwap_trade_details):
    # Metrics for RL model
    rl_metrics = {
        'slippage':total_slippage,
        'price_impact': sum(env.data['transaction_cost'].iloc[trade['step']] * trade['Volume'] for trade in rl_trade_details),
        'spread_cost': sum(trade['Spread Cost'] for trade in rl_trade_details),
        'opportunity_cost': env.start_trade_price - env.current_average_trade_price
    }

    def get_prices_for_step(trade, env):
        timestamp_str = env.data['Timestamp'].iloc[int(trade['step'])]
        ticker_symbol = env.data['Ticker'].iloc[int(trade['step'])]
        expected_price, bid_sizes, ask_sizes, bid_prices, ask_prices = env.get_expected_price(ticker_symbol, timestamp_str)
        
        # Ensure these are treated as lists
        if not isinstance(bid_prices, list):
            bid_prices = [bid_prices]
        if not isinstance(bid_sizes, list):
            bid_sizes = [bid_sizes]
        if not isinstance(ask_prices, list):
            ask_prices = [ask_prices]
        if not isinstance(ask_sizes, list):
            ask_sizes = [ask_sizes]

        return bid_prices, ask_prices, expected_price, bid_sizes

    # Metrics for TWAP strategy
    twap_slippage = sum(
        abs(max(get_prices_for_step(trade, env)[0]) - trade['execution_price']) * trade['shares']
        for _, trade in twap_trade_details.iterrows()
    )
    twap_price_impact = sum(
        env.data['transaction_cost'].iloc[int(trade['step'])] * trade['shares']
        for _, trade in twap_trade_details.iterrows()
    )
    twap_spread_cost = sum(
        (min(get_prices_for_step(trade, env)[1]) - get_prices_for_step(trade, env)[2]) * trade['shares']
        for _, trade in twap_trade_details.iterrows()
    )

    twap_metrics = {
        'slippage': twap_slippage,
        'price_impact': twap_price_impact,
        'spread_cost': twap_spread_cost,
        'opportunity_cost': np.mean([
            env.start_trade_price - trade['execution_price']
            for _, trade in twap_trade_details.iterrows()
        ])
    }

    # Metrics for VWAP strategy
    vwap_slippage = sum(
        abs(max(get_prices_for_step(trade, env)[0]) - trade['execution_price']) * trade['shares']
        for _, trade in vwap_trade_details.iterrows()
    )
    vwap_price_impact = sum(
        env.data['transaction_cost'].iloc[int(trade['step'])] * trade['shares']
        for _, trade in vwap_trade_details.iterrows()
    )
    vwap_spread_cost = sum(
        (min(get_prices_for_step(trade, env)[1]) - get_prices_for_step(trade, env)[2]) * trade['shares']
        for _, trade in vwap_trade_details.iterrows()
    )

    vwap_metrics = {
        'slippage': vwap_slippage,
        'price_impact': vwap_price_impact,
        'spread_cost': vwap_spread_cost,
        'opportunity_cost': np.mean([
            env.start_trade_price - trade['execution_price']
            for _, trade in vwap_trade_details.iterrows()
        ])
    }

    return rl_metrics, twap_metrics, vwap_metrics
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Visualization Function
def plot_comparison_metrics(rl_metrics, twap_metrics, vwap_metrics):
    labels = ['Slippage', 'Price Impact', 'Spread Cost', 'Opportunity Cost']
    rl_values = [rl_metrics['slippage'], rl_metrics['price_impact'], rl_metrics['spread_cost'], rl_metrics['opportunity_cost']]
    twap_values = [twap_metrics['slippage'], twap_metrics['price_impact'], twap_metrics['spread_cost'], twap_metrics['opportunity_cost']]
    vwap_values = [vwap_metrics['slippage'], vwap_metrics['price_impact'], vwap_metrics['spread_cost'], vwap_metrics['opportunity_cost']]

    # Ensure all values are scalar
    def sanitize_value(value):
        if isinstance(value, (list, np.ndarray)):
            value = np.mean(value)
        return value  # Return the original value without applying log

    rl_values = [sanitize_value(value) for value in rl_values]
    twap_values = [sanitize_value(value) for value in twap_values]
    vwap_values = [sanitize_value(value) for value in vwap_values]

    # Plotting each metric separately
    for i, label in enumerate(labels):
        plt.figure(figsize=(8, 5))
        plt.bar(['RL Model', 'TWAP', 'VWAP'], [rl_values[i], twap_values[i], vwap_values[i]], color=['blue', 'orange', 'green'])
        plt.xlabel('Strategies')
        plt.ylabel(f'{label}')
        plt.title(f'Comparison of {label} Across Strategies')
        plt.show()

    # Print the values along with metric names to inspect them
    print("RL Model Metrics:")
    for label, value in zip(labels, rl_values):
        print(f"{label}: {value}")
    
    print("\nTWAP Model Metrics:")
    for label, value in zip(labels, twap_values):
        print(f"{label}: {value}")
    
    print("\nVWAP Model Metrics:")
    for label, value in zip(labels, vwap_values):
        print(f"{label}: {value}")
```

```python
import os
import time
from datetime import datetime, timedelta

import pandas as pd
import numpy as np
import torch

import warnings

from blockhouse_ml.utils.macro_model import MetaLearner
from blockhouse_ml.utils.data_handler import DataProcessor, InferenceDataHandler
from blockhouse_ml.utils.macro_model_utils import MacroTraderModel
from blockhouse_ml.utils import fetch_merge_data
from blockhouse_ml.utils.env import CustomTradingEnvironment, TradingEnvironmentMicro
from blockhouse_ml.utils.micro_model import TabTransformerPolicy, TabTransformerExtractor
from blockhouse_ml.utils.micro_model_utils import MicroTraderModel
# from blockhouse_ml.utils.micro_model_utils import MicroTraderModel

os.environ['KMP_DUPLICATE_LIB_OK'] = 'TRUE'
os.environ['OMP_NUM_THREADS'] = '1'
'''
Because of this error:
 [error] Disposing session as kernel process died ExitCode: 3, Reason: OMP: Error #15: Initializing libiomp5md.dll, but found libomp140.x86_64.dll already initialized.
OMP: Hint This means that multiple copies of the OpenMP runtime have been linked into the program. That is dangerous, since it can degrade performance or cause incorrect results. The best thing to do is to ensure that only a single OpenMP runtime is linked into the process, e.g. by avoiding static linking of the OpenMP runtime in any library. As an unsafe, unsupported, undocumented workaround you can set the environment variable KMP_DUPLICATE_LIB_OK=TRUE to allow the program to continue to execute, but that may cause crashes or silently produce incorrect results. For more information, please see http://www.intel.com/software/products/support/.
'''

warnings.filterwarnings("ignore")

## Initialize the variables
# Initialize the model directory, where the models will be saved
MACRO_MODEL_DIR = 'TabModels'
os.makedirs(MACRO_MODEL_DIR, exist_ok=True)

MICRO_MODEL_DIR = 'MicroModels'
os.makedirs(MICRO_MODEL_DIR, exist_ok=True)
micro_trader = MicroTraderModel()

# Initialize the data directory, where the data will be stored
data_dir = 'Data'
os.makedirs(data_dir, exist_ok=True)

# Initialize the MacroTraderModel
macro_trader = MacroTraderModel(MACRO_MODEL_DIR)

# Initialize the MicroTraderModel
# micro_trader = MicroTraderModel(MICRO_MODEL_DIR)

# Initialize the data processor
data_processor = DataProcessor()

# Inference data handler
inference_data_handler = InferenceDataHandler()

# Initialize the MetaLearner
meta = MetaLearner()


def get_schedule(timeframe, transaction_size, market_cap_int, input_row, data):
    """
    Generates a trading schedule based on the transaction size and input data.

    Args:
    - timeframe (int): The timeframe for trade execution.
    - transaction_size (int): The size of the transaction.
    - input_row (pd.DataFrame): The input data row with forecasts and technical indicators.

    Returns:
    - list: A list of trades executed based on the generated schedule.
    """
    print("LOGGING: Generating Schedule...")
    trades, micro_input = macro_trader.infer_macro(timeframe, transaction_size, market_cap_int,  input_row, data, meta=meta, inference_data_handler=inference_data_handler)
    return trades, micro_input

def run_pipeline(ticker, start_timestamp = (datetime.today() - timedelta(days=7)).strftime('%Y-%m-%d'), end_timestamp=datetime.today().strftime('%Y-%m-%d'), timeframe=390, inventory=10000, trade_set_counter=1):
    """
    Runs the entire pipeline for the trading model, including data retrieval, 
    technical indicator addition, forecasting, and generating trade schedules.

    Returns:
    - list: The list of trades generated by the model.
    """
    api_key = 'r65B9O5aplSJn7BWSo8z8pNH8v2wW5yc'
    start_time = time.time()
    # Create the trading environment
    data = fetch_merge_data.fetch_and_merge_data(ticker,start_date=start_timestamp,end_date=end_timestamp,save_dir =data_dir)
    data = data_processor.add_technical_indicators(data)
    input_row = inference_data_handler.add_forecasts(data)
    market_cap_int = fetch_merge_data.get_market_cap(ticker)
    
    trades, micro_input = get_schedule(timeframe, inventory, market_cap_int, input_row, data)
    trades.to_csv("tradessss.csv", index=False)
    # Inspect the columns in the trades DataFrame
    print("Columns in 'trades' DataFrame:", trades.columns)
    
    # Assuming 'timestamp' is the correct column name, check if it exists
    if 'timestamp' in trades.columns:
        micro_input['Timestamp'] = trades['timestamp'].tolist()
    else:
        print("Error: 'timestamp' column not found in 'trades' DataFrame")
        return
    
    if isinstance(micro_input['Timestamp'].iloc[0], str) and "Name:" in micro_input['Timestamp'].iloc[0]:
        # Extract just the date and time from the first row
        clean_timestamp = micro_input['Timestamp'].iloc[0].split('Name:')[0].strip()
        micro_input['Timestamp'].iloc[0] = clean_timestamp

    # Ensure the 'Timestamp' column is in proper datetime format for all rows
    micro_input['Timestamp'] = pd.to_datetime(micro_input['Timestamp'], errors='ignore')
    
    # Add additional columns to micro_input
    micro_input['Ticker'] = [ticker] * len(micro_input)
    micro_input['Inventory'] = [inventory] * len(micro_input)
    micro_input['Trade_Set_ID'] = [f"set_{trade_set_counter}"] * len(micro_input)
    trades = pd.DataFrame(trades)
    micro_input['shares'] = trades['shares'].values


    # ## run micro trader
    micro_env = TradingEnvironmentMicro(micro_input, api_key, preferred_timeframe=timeframe, target_shares=inventory)
    policy_kwargs = {'state_cols': micro_env.state_columns}
    micro_model = micro_trader.load_model(micro_env, policy_kwargs) 
 
    # Reset the environment to start inference
    obs = micro_env.reset()

    # Initialize an empty list to store trade details
    trade_details = []

    for _ in range(len(micro_input)):
        # Predict the action to take based on the current observation
        action, _states = micro_model.predict(obs)
        
        # Step through the micro_environment using the predicted action
        obs, rewards, done, info = micro_env.step(action)
        
        # # Extract and print the trade details
        # order_type = "Market" if action[0] < 0.5 else "Limit"
        # execution_price = action[1] if order_type == "Limit" else obs[3]  # Assuming 'close' price is at index 3
        
        # # Append the trade details to the list
        # trade_details.append({
        #     'Order Type': order_type,
        #     'Limit Price': execution_price
        # })
        
        # print(f"Trade Executed: {order_type} Order at Price: {execution_price}")
        
        # If the micro_environment is done, break the loop
        if done:
            break
    trade_details = micro_env.results
    # Render the final state of the micro_environment
    micro_env.render()
    
    return trade_details
    # return trades
```
