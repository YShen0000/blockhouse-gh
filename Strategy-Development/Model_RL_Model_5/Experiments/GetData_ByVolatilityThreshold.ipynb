```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import random
```

```python
csv_path = 'merged_bid_ask_ohlcv_data_2.csv'
df = pd.read_csv(csv_path)
df.dropna(subset=['volatility'], inplace=True)
df.head()
```

```python
df.hist(column='volatility', bins=10, color='c', edgecolor='k', alpha=0.65)
# Add a title and labels
plt.title('Histogram')
plt.xlabel('Volatility')
plt.ylabel('Frequency')

# Show the plot
plt.show()
```
