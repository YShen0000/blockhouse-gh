## This NB is for deploying the RL model to production, i.e, to create a Sagemaker endpoint for this model

### Create tarball archive of the model files

```python
import tarfile
import os

def create_model_package(model_files, output_path):
    with tarfile.open(output_path, "w:gz") as tar:
        for file in model_files:
            print(os.path.basename(file))
            tar.add(file, arcname=os.path.join("code", os.path.basename(file)))
        # tar.add(inference_script)# arcname="code/inference.py")

# Usage
model_files = ["../backtest_lib","inference.py", "requirements.txt"]
output_path = "Statmodel_version_2.tar.gz"
create_model_package(model_files, output_path)
```
