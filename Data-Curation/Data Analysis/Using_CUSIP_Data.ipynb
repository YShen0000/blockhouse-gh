```python
import pandas as pd
```

```python
data1 = pd.read_csv("Database for CUSIP of Treasury/Securities_1.csv")
data2 = pd.read_csv("Database for CUSIP of Treasury/Securities_2.csv")
data3 = pd.read_csv("Database for CUSIP of Treasury/Securities_3.csv")
data4 = pd.read_csv("Database for CUSIP of Treasury/Securities_4.csv")
data5 = pd.read_csv("Database for CUSIP of Treasury/Securities_5.csv")
data6 = pd.read_csv("Database for CUSIP of Treasury/Securities_6.csv")

securities_df = pd.concat([data1, data2, data3, data4, data5, data6], axis =0)
securities_df.head()
```

```python
len(securities_df)
```

```python
def find_issue_date(cusip, dataframe):
    # Search for the CUSIP in the DataFrame
    record = dataframe[dataframe['CUSIP'] == cusip]
    
    # Check if the record is found
    if not record.empty:
        issue_date = record.iloc[0]['Issue Date']
        return issue_date
    else:
        return f"No record found for CUSIP {cusip}."

# Example usage
cusip_input = "912828Y87"  # Replace with the CUSIP you want to search
find_issue_date(cusip_input, securities_df)
```

```python
# Specify the path where you want to save the Excel file
output_file = "Securities_Date_10_02_2024.xlsx"

# Export the DataFrame to Excel
securities_df.to_excel(output_file, index=False)
print(f"Data has been successfully exported to {output_file}")
```
