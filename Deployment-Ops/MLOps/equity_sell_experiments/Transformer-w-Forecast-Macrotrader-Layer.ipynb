### Calculating AC values

```python
import numpy as np
import math

# Almgren-Chriss model parameters
x0 = 5000  # Initial number of shares to trade
T = 1.0    # Total time horizon (e.g., 1 day)
N = min(x0, 2400)  # Number of discrete time intervals
eta = 0.0000001    # Temporary/permanent impact coefficient
sigma = 0.02       # Volatility of the asset
lambda_ = 0.1      # Risk aversion parameter

# Time interval
dt = 1

# Cost function for the Almgren-Chriss model
def cost_function(x, eta, sigma, x0=10):
    x_cumsum = np.cumsum(x)
    x_half = x / 2

    temp_cost = np.sum(eta * (x**2) / dt)
    perm_cost = np.sum(eta * x * (x0 - x_cumsum + x_half))
    var_cost = np.sum(lambda_ * sigma**2 * (x**2) * dt)

    return (1 / (temp_cost + perm_cost + var_cost) / x0) * 10**(math.log10(x0) * 4 - 6)

def almgren(row):
    x0 = 5000
    N = min(x0, 2400)

    eta = 1 / row['mean_liq'] if row['mean_liq'] != 0 else 0.0000001
    sigma = row['mean_vol']

    # Initial guess: trade evenly across all intervals
    x_init = np.ones(N) * (x0 / N)
    
    return cost_function(x_init, eta, sigma, x0) / x0

# Apply the almgren function to each row in the DataFrame
data['transaction_cost'] = data.apply(almgren, axis=1)
```

```python
# Function to resample and group by ticker
def resample_and_group(data, freq):
    # Convert timestamp to datetime
    data['timestamp'] = pd.to_datetime(data['timestamp'])
    resampled_data = data.set_index('timestamp').groupby('symbol').resample(freq).agg({
        'open': 'mean',
        'high': 'mean',
        'low': 'mean',
        'close': 'mean',
        'volume': 'sum',
        'transaction_cost': 'mean',  # Update this to your TC column name if different
        'bid_px_00': 'last',
        'ask_px_00': 'last',
        'bid_sz_00': 'sum',
        'ask_sz_00': 'sum',
        'price': 'last',
        'size': 'sum'
    }).reset_index()
    return resampled_data

# Create daily, weekly, and monthly datasets
daily_data = resample_and_group(data, 'D')
# weekly_data = resample_and_group(data, 'W')
# monthly_data = resample_and_group(data, 'M')

daily_data = daily_data.dropna()
# weekly_data = weekly_data.dropna()
# monthly_data = monthly_data.dropna()
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

# Assumption: We have transaction cost in column
def create_datasets(data, max_encoder_length=80, max_prediction_length=80):
    # Convert timestamp to datetime
    data['timestamp'] = pd.to_datetime(data['timestamp'])

    # Ensure 'time_idx' is a column representing the order of data points
    data['time_idx'] = data.index

    # Create time-based features
    data['day_of_week'] = data['timestamp'].dt.dayofweek.astype(str)
    data['hour'] = data['timestamp'].dt.hour.astype(str)

    # Normalize the numerical features
    scaler = StandardScaler()
    data[['open', 'high', 'low', 'close', 'volume', 'bid_px_00',  'ask_px_00', 'ask_sz_00', 'bid_sz_00']] = scaler.fit_transform(
        data[['open', 'high', 'low', 'close', 'volume', 'bid_px_00', 'ask_px_00', 'ask_sz_00', 'bid_sz_00']]
    )

    # Ensure 'symbol' is treated as a categorical variable
    data['symbol'] = data['symbol'].astype(str).astype('category')

    # Define the TFT Dataset
    from pytorch_forecasting import TimeSeriesDataSet

    dataset = TimeSeriesDataSet(
        data,
        time_idx="time_idx",
        target="transaction_cost",  # Assuming we have transaction_cost in columns
        group_ids=["symbol"],
        min_encoder_length=max_encoder_length // 2,
        max_encoder_length=max_encoder_length,
        min_prediction_length=1,
        max_prediction_length=max_prediction_length,
        static_categoricals=["symbol"],
        time_varying_known_categoricals=["day_of_week", "hour"],
        time_varying_known_reals=["time_idx"],
        time_varying_unknown_reals=["open", "high", "low", "close", "volume", "bid_px_00", "ask_px_00", "ask_sz_00", "bid_sz_00"],
        add_relative_time_idx=True,
        add_target_scales=True,
        add_encoder_length=True,
        allow_missing_timesteps=True
    )
    print(type(dataset))
    # Split data into training and validation sets
    # Define the indices for training and validation sets
    train_size = int(len(data) * 0.8)
    train_data = data.iloc[:train_size]
    val_data = data.iloc[train_size:]

    train_dataset = TimeSeriesDataSet.from_dataset(dataset, train_data, predict=False)
    val_dataset = TimeSeriesDataSet.from_dataset(dataset, val_data, predict=False)

    return train_dataset, val_dataset
```

### Training the TFT Model

```python
import pytorch_lightning as pl
import torch
import warnings
warnings.filterwarnings("ignore")
from pytorch_forecasting.metrics import MAE

# Create dataloaders
batch_size = 64

# Testing daily data
train_dataset, val_dataset = create_datasets(daily_data)
train_dataloader = train_dataset.to_dataloader(train=True, batch_size=batch_size, num_workers=0)
val_dataloader = val_dataset.to_dataloader(train=False, batch_size=batch_size, num_workers=0)

# Model setup and config
from pytorch_forecasting import TemporalFusionTransformer
from pytorch_forecasting import TimeSeriesDataSet, TemporalFusionTransformer, QuantileLoss
import numpy as np

# Define the TFT model
tft = TemporalFusionTransformer.from_dataset(
    train_dataset,
    learning_rate=0.09,
    hidden_size=16,
    attention_head_size=8,
    dropout=0.1,
    hidden_continuous_size=8,
    output_size=7,  # 7 quantiles by default
    loss=QuantileLoss(),
    log_interval=10,
    reduce_on_plateau_patience=4,
)



class TFTLightningModule(pl.LightningModule):
    def __init__(self, tft):
        super().__init__()
        self.tft = tft
        self.loss = QuantileLoss()
        self.mae = MAE()
        self.train_losses = []
        self.val_losses = []
        self.mae_train_losses = []
        self.mae_val_losses = []

    def forward(self, x):
        return self.tft(x)

    def training_step(self, batch, batch_idx):
        x, y = batch
        output = self.tft(x)
        loss = self.loss(output["prediction"], y[0])

        # Calculate MAE for each forecast horizon
        for t in range(output["prediction"].size(1)):
            mae = torch.nn.functional.l1_loss(output["prediction"][:, t, 0], y[0][:, t])
            self.log(f'train_mae_step_{t}', mae, prog_bar=True, logger=True)
            self.mae_train_losses.append(mae.item())
            
#         self.mae_train_losses.append(mae.item())
        self.train_losses.append(loss.item())
        return loss

    def validation_step(self, batch, batch_idx):
        x, y = batch
        output = self.tft(x)
        val_loss = self.loss(output["prediction"], y[0])

        # Calculate MAE for each forecast horizon
        for t in range(output["prediction"].size(1)):
            val_mae = torch.nn.functional.l1_loss(output["prediction"][:, t, 0], y[0][:, t])
            self.log(f'val_mae_step_{t}', val_mae, prog_bar=True, logger=True)
            self.mae_val_losses.append(val_mae.item())

#         self.mae_val_losses.append(val_mae.item())
        self.val_losses.append(val_loss.item())
        return val_loss
    
    def on_train_epoch_end(self):
        avg_train_loss = torch.tensor(self.train_losses).mean().item()
        avg_mae_train_loss = torch.tensor(self.mae_train_losses).mean().item()
        self.train_losses = []  # Reset for the next epoch
        self.mae_train_losses = []
        print(f'Epoch {self.current_epoch}, Average Training Loss: {avg_train_loss}, Average MAE Train Loss: {avg_mae_train_loss}')

    def on_validation_epoch_end(self):
        avg_val_loss = torch.tensor(self.val_losses).mean().item()
        avg_mae_val_loss = torch.tensor(self.mae_val_losses).mean().item()
        self.val_losses = []  # Reset for the next epoch
        self.mae_val_losses = []
        print(f'Epoch {self.current_epoch}, Average Validation Loss: {avg_val_loss}, Average MAE Val Loss: {avg_mae_val_loss}')

    def configure_optimizers(self):
        return torch.optim.Adam(self.parameters(), lr=0.03)

    
# ############################################################
    
# # Debug function to check batch data
# def debug_batch(dataloader, num_batches=1):
#     for i, batch in enumerate(dataloader):
#         if i >= num_batches:
#             break
#         print(f"Batch {i + 1}/{num_batches}")
#         x, y = batch
#         for key, value in x.items():
#             print(f"{key}: {value.shape}")

# # Debug the first few batches to verify the format
# print("Training DataLoader:")
# debug_batch(train_dataloader)

# print("\nValidation DataLoader:")
# debug_batch(val_dataloader)


# ############################################################

# Instantiate the wrapped model
tft_lightning = TFTLightningModule(tft)

# Ensure that tft is recognized as a LightningModule
assert isinstance(tft_lightning, pl.LightningModule)

# Define checkpoint callback to save the model
checkpoint_callback = pl.callbacks.ModelCheckpoint(
    dirpath="checkpoints",  # Save to the 'checkpoints' directory locally
    filename="tft-{epoch:02d}-{val_loss:.2f}",
    save_top_k=1,
    monitor="val_loss",
    mode="min"
)


# Train the model
trainer = pl.Trainer(
    max_epochs=10,
    gradient_clip_val=0.1,
)
trainer.fit(
    tft_lightning,
    train_dataloaders=train_dataloader,
    val_dataloaders=val_dataloader,
)
```

```python
import pandas as pd
import numpy as np
import talib as ta

class TechnicalIndicators:
    def __init__(self, data):
        self.data = data

    def add_momentum_indicators(self):
        self.data['RSI'] = ta.RSI(self.data['close'], timeperiod=14)
        self.data['MACD'], self.data['MACD_signal'], self.data['MACD_hist'] = ta.MACD(self.data['close'], fastperiod=12, slowperiod=26, signalperiod=9)
        self.data['Stoch_k'], self.data['Stoch_d'] = ta.STOCH(self.data['high'], self.data['low'], self.data['close'],
                                                              fastk_period=14, slowk_period=3, slowd_period=3)

    def add_volume_indicators(self):
        self.data['OBV'] = ta.OBV(self.data['close'], self.data['volume'])

    def add_volatility_indicators(self):
        self.data['Upper_BB'], self.data['Middle_BB'], self.data['Lower_BB'] = ta.BBANDS(self.data['close'], timeperiod=20)
        self.data['ATR_1'] = ta.ATR(self.data['high'], self.data['low'], self.data['close'], timeperiod=1)
        self.data['ATR_2'] = ta.ATR(self.data['high'], self.data['low'], self.data['close'], timeperiod=2)
        self.data['ATR_5'] = ta.ATR(self.data['high'], self.data['low'], self.data['close'], timeperiod=5)
        self.data['ATR_10'] = ta.ATR(self.data['high'], self.data['low'], self.data['close'], timeperiod=10)
        self.data['ATR_20'] = ta.ATR(self.data['high'], self.data['low'], self.data['close'], timeperiod=20)
        
    # Adding volatility to calculate dynamic volatility
    def add_volatility(self, window=15):
        self.data['log_return'] = np.log(self.data['close'] / self.data['close'].shift(1))
        self.data['volatility'] = self.data['log_return'].rolling(window=window).std()*np.sqrt(window)

    def add_trend_indicators(self):
        self.data['ADX'] = ta.ADX(self.data['high'], self.data['low'], self.data['close'], timeperiod=14)
        self.data['+DI'] = ta.PLUS_DI(self.data['high'], self.data['low'], self.data['close'], timeperiod=14)
        self.data['-DI'] = ta.MINUS_DI(self.data['high'], self.data['low'], self.data['close'], timeperiod=14)
        self.data['CCI'] = ta.CCI(self.data['high'], self.data['low'], self.data['close'], timeperiod=5)

    def add_other_indicators(self):
        self.data['DLR'] = np.log(self.data['close'] / self.data['close'].shift(1))
        self.data['TWAP'] = self.data['close'].expanding().mean()
        self.data['VWAP'] = (self.data['volume'] * (self.data['high'] + self.data['low']) / 2).cumsum() / self.data['volume'].cumsum()
        self.data['market_liquidity'] = self.data['bid_sz_00'] + self.data['ask_sz_00']
        self.data['expected_price'] = self.data['ask_px_00']

    def add_all_indicators(self):
        self.add_momentum_indicators()
        self.add_volume_indicators()
        self.add_volatility_indicators()
        self.add_trend_indicators()
        self.add_other_indicators()
        self.add_volatility()
        return self.data
```

### 1. Making forecasts for days, weeks and month

```python
import torch
from pytorch_forecasting import TemporalFusionTransformer
from pytorch_forecasting.data import TimeSeriesDataSet
from sklearn.preprocessing import StandardScaler
import pandas as pd

class Forecast_TC:
    def __init__(self, data, model):
        self.data = data
        self.model = model
    
    def preprocess_data(self):
        # Convert timestamp to datetime
        self.data['timestamp'] = pd.to_datetime(self.data['timestamp'])

        # Ensure 'time_idx' is a column representing the order of data points
        self.data['time_idx'] = self.data.index
        
        self.data['symbol'] = self.data['symbol'].astype(str).astype('category')

        # Create time-based features
        self.data['day_of_week'] = self.data['timestamp'].dt.dayofweek.astype(str)
        self.data['hour'] = self.data['timestamp'].dt.hour.astype(str)

        # Normalize the numerical features
        scaler = StandardScaler()
        self.data[['open', 'high', 'low', 'close', 'volume', 'bid_px_00',  'ask_px_00', 'ask_sz_00', 'bid_sz_00']] = scaler.fit_transform(
            self.data[['open', 'high', 'low', 'close', 'volume', 'bid_px_00', 'ask_px_00', 'ask_sz_00', 'bid_sz_00']]
        )
        
        # Ensure all relevant columns are categorical
        self.data['day_of_week'] = self.data['day_of_week'].astype('category')
        self.data['hour'] = self.data['hour'].astype('category')
     
        return self.data

    def make_forecasts(self, max_encoder_length=60, max_prediction_length=60, hours = False):
#         max_encoder_length = 60  # Length of historical sequence
#         max_prediction_length = 60  # Length of forecast horizon ( 1month)
        data = self.preprocess_data() # Get data preprocessed
        model = self.model
        print(len(data))
        dataset = TimeSeriesDataSet(
            data,
            time_idx="time_idx",
            target="transaction_cost",  # Update this to your TC column name if different
            group_ids=["symbol"],
            min_encoder_length=max_encoder_length // 2,
            max_encoder_length=max_encoder_length,
            min_prediction_length=1,
            max_prediction_length=max_prediction_length,
            static_categoricals=["symbol"],
            time_varying_known_categoricals=["day_of_week", "hour"],
            time_varying_known_reals=["time_idx"],
            time_varying_unknown_reals=["open", "high", "low", "close", "volume", "bid_px_00", "ask_px_00", "ask_sz_00", "bid_sz_00"],
            add_relative_time_idx=True,
            add_target_scales=True,
            add_encoder_length=True,
            allow_missing_timesteps=True
        )

        # Create dataloader
        dataloader = dataset.to_dataloader(train=False, batch_size=1, num_workers=0)
        print(len(dataloader))
        
        if hours == False:
            # Initialize lists to store forecasts
            forecast_1_day_list = []
            forecast_1_week_list = []
            forecast_2_weeks_list = []
            forecast_1_month_list = []

            # Get the prediction for each row
            with torch.no_grad():
                for batch in dataloader:
                    x, y = batch
                    prediction = model(x)
    #                 print(prediction['prediction'].shape)
                    # Extract predictions for specific future steps
                    forecast_1_day = prediction['prediction'][0, 1, 0].item() if prediction['prediction'].shape[1] > 1 else None
                    forecast_1_week = prediction['prediction'][0, 7, 0].item() if prediction['prediction'].shape[1] > 7 else None
                    forecast_2_weeks = prediction['prediction'][0, 14, 0].item() if prediction['prediction'].shape[1] > 14 else None
                    forecast_1_month = prediction['prediction'][0, 29, 0].item() if prediction['prediction'].shape[1] > 29 else None

                    # Append the forecasts to the lists
                    forecast_1_day_list.append(forecast_1_day)
                    forecast_1_week_list.append(forecast_1_week)
                    forecast_2_weeks_list.append(forecast_2_weeks)
                    forecast_1_month_list.append(forecast_1_month)

            # Debugging information
            print(f"Processed data length: {len(data)}")
            print(f"Forecast 1 day list length: {len(forecast_1_day_list)}")
            print(f"Forecast 1 week list length: {len(forecast_1_week_list)}")
            print(f"Forecast 2 weeks list length: {len(forecast_2_weeks_list)}")
            print(f"Forecast 1 month list length: {len(forecast_1_month_list)}")

            # Ensure the forecast lists have the same length as the original data
            if len(forecast_1_day_list) != len(data):
                print("Length mismatch found, adjusting forecast lists to match data length.")
            forecast_1_day_list = forecast_1_day_list[:len(data)]
            forecast_1_week_list = forecast_1_week_list[:len(data)]
            forecast_2_weeks_list = forecast_2_weeks_list[:len(data)]
            forecast_1_month_list = forecast_1_month_list[:len(data)]

            # Add the forecasts to the original data
            data['forecast_1_day'] = forecast_1_day_list
            data['forecast_1_week'] = forecast_1_week_list
            data['forecast_2_weeks'] = forecast_2_weeks_list
            data['forecast_1_month'] = forecast_1_month_list
        
        else:
            # Make predictions for hourly data
            
            # Initialize lists to store forecasts
            forecast_1_hour_list = []
            forecast_2_hours_list = []
            forecast_3_hours_list = []
            
            # Get the prediction for each row
            with torch.no_grad():
                for batch in dataloader:
                    x, y = batch
                    prediction = model(x)
                    # Extract predictions for specific future steps
                    forecast_1_hour = prediction['prediction'][0, 59, 0].item() if prediction['prediction'].shape[1] > 59 else None
                    forecast_2_hours = prediction['prediction'][0, 119, 0].item() if prediction['prediction'].shape[1] > 119 else None
                    forecast_3_hours = prediction['prediction'][0, 179, 0].item() if prediction['prediction'].shape[1] > 179 else None

                    # Append the forecasts to the lists
                    forecast_1_hour_list.append(forecast_1_hour)
                    forecast_2_hours_list.append(forecast_2_hours)
                    forecast_3_hours_list.append(forecast_3_hours)

            # Debugging information
            print(f"Processed data length: {len(data)}")
            print(f"Forecast 1 hour list length: {len(forecast_1_hour_list)}")
            print(f"Forecast 2 hours list length: {len(forecast_2_hours_list)}")
            print(f"Forecast 3 hours list length: {len(forecast_3_hours_list)}")

            # Ensure the forecast lists have the same length as the original data
            if len(forecast_1_hour_list) != len(data):
                print("Length mismatch found, adjusting forecast lists to match data length.")
            forecast_1_hour_list = forecast_1_hour_list[:len(data)]
            forecast_2_hours_list = forecast_2_hours_list[:len(data)]
            forecast_3_hours_list = forecast_3_hours_list[:len(data)]

            # Add the forecasts to the original data
            data['forecast_1_hour'] = forecast_1_hour_list
            data['forecast_2_hours'] = forecast_2_hours_list
            data['forecast_3_hours'] = forecast_3_hours_list
        
        
            

        return data
```

```python
import pandas as pd
import numpy as np
import talib as ta

class TechnicalIndicators:
    def __init__(self, data):
        self.data = data

    def add_momentum_indicators(self):
        self.data['RSI'] = ta.RSI(self.data['close'], timeperiod=14)
        self.data['MACD'], self.data['MACD_signal'], self.data['MACD_hist'] = ta.MACD(self.data['close'], fastperiod=12, slowperiod=26, signalperiod=9)
        self.data['Stoch_k'], self.data['Stoch_d'] = ta.STOCH(self.data['high'], self.data['low'], self.data['close'],
                                                              fastk_period=14, slowk_period=3, slowd_period=3)

    def add_volume_indicators(self):
        self.data['OBV'] = ta.OBV(self.data['close'], self.data['volume'])

    def add_volatility_indicators(self):
        self.data['Upper_BB'], self.data['Middle_BB'], self.data['Lower_BB'] = ta.BBANDS(self.data['close'], timeperiod=20)
        self.data['ATR_1'] = ta.ATR(self.data['high'], self.data['low'], self.data['close'], timeperiod=1)
        self.data['ATR_2'] = ta.ATR(self.data['high'], self.data['low'], self.data['close'], timeperiod=2)
        self.data['ATR_5'] = ta.ATR(self.data['high'], self.data['low'], self.data['close'], timeperiod=5)
        self.data['ATR_10'] = ta.ATR(self.data['high'], self.data['low'], self.data['close'], timeperiod=10)
        self.data['ATR_20'] = ta.ATR(self.data['high'], self.data['low'], self.data['close'], timeperiod=20)
        
    # Adding volatility to calculate dynamic volatility
    def add_volatility(self, window=15):
        self.data['log_return'] = np.log(self.data['close'] / self.data['close'].shift(1))
        self.data['volatility'] = self.data['log_return'].rolling(window=window).std()*np.sqrt(window)

    def add_trend_indicators(self):
        self.data['ADX'] = ta.ADX(self.data['high'], self.data['low'], self.data['close'], timeperiod=14)
        self.data['+DI'] = ta.PLUS_DI(self.data['high'], self.data['low'], self.data['close'], timeperiod=14)
        self.data['-DI'] = ta.MINUS_DI(self.data['high'], self.data['low'], self.data['close'], timeperiod=14)
        self.data['CCI'] = ta.CCI(self.data['high'], self.data['low'], self.data['close'], timeperiod=5)

    def add_other_indicators(self):
        self.data['DLR'] = np.log(self.data['close'] / self.data['close'].shift(1))
        self.data['TWAP'] = self.data['close'].expanding().mean()
        self.data['VWAP'] = (self.data['volume'] * (self.data['high'] + self.data['low']) / 2).cumsum() / self.data['volume'].cumsum()
        self.data['market_liquidity'] = self.data['bid_sz_00'] + self.data['ask_sz_00']
        self.data['expected_price'] = self.data['ask_px_00']

    def add_all_indicators(self):
        self.add_momentum_indicators()
        self.add_volume_indicators()
        self.add_volatility_indicators()
        self.add_trend_indicators()
        self.add_other_indicators()
        self.add_volatility()
        return self.data
```

```python
# To do: Add market liquidity, TWAP, VWAP columns to data
# Hyperparameters: Increased entropy, Added noise, Tested alpha beta gamma values, Learning Rates, etc

import gym
from gym import spaces
import numpy as np
import pandas as pd

class TradingEnvironment(gym.Env):
    metadata = {'render.modes': ['human']}
    
    # Preferred timeframe: The number of steps that the user wants to complete the trade in
    def __init__(self, data,scenario, daily_trading_limit=10000, preferred_timeframe=10000, initial_inventory=1000):
        super(TradingEnvironment, self).__init__()
        self.data = data
        self.daily_trading_limit = daily_trading_limit
        self.current_step = 0
        
        self.preferred_timeframe = preferred_timeframe
        self.initial_inventory = initial_inventory
        self.remaining_inventory = self.initial_inventory
        self.elapsed_time = 0
        self.trades = []
        self.cumulative_reward = 0
        
        # Define scenario-specific penalties
        if scenario == 'high_frequency':
            self.beta = 10
            self.delta = 10
        elif scenario == 'low_frequency':
            self.beta = 1e5
            self.delta = 1e5
        else:  # medium_frequency
            self.beta = 1e3
            self.delta = 1e3

        # Extract state columns - have to change according to requirement
        self.state_columns = ['close', 'volume', 'RSI', 'MACD', 'MACD_signal', 'MACD_hist', 'Stoch_k', 'Stoch_d',
                              'OBV', 'Upper_BB', 'Middle_BB', 'Lower_BB', 'ATR_1', 'ADX', '+DI', '-DI', 'CCI']
        
        # Define action space - [%_slice, timing_of_next_slice]
        # Here the upper limit for time is set to 10 but can be changed if necessary
        self.action_space = spaces.Box(low=np.array([0.01, 1]), high=np.array([0.20, 100]), dtype=np.float32)

        # Define observation space based on state columns + 1 (remaining_inventory)
        self.observation_space = spaces.Box(low=-np.inf, high=np.inf, shape=(len(self.state_columns) + 1,), dtype=np.float32)
        
        # Calculate historical volatility
#         self.data['Volatility'] = calculate_historical_volatility(self.data)
    
    def _get_state(self):
        market_conditions = self._next_observation()
        state = np.append(market_conditions, self.remaining_inventory / self.initial_inventory)
        return state
    
    def _next_observation(self):
        return self.data[self.state_columns].iloc[self.current_step].values
    
    def reset(self):
        print('!!!!!!!!!!Class resetted!!!!!!!!!!!!!!')
        self.current_step = 0
        self.cumulative_reward = 0
        self.remaining_inventory = self.initial_inventory
        self.elapsed_time = 0
        self.trades = []
        return self._get_state()
    
    def step(self, action):
        # Adding some noise to the actions
        action = self._add_noise_to_action(action)
        print(f'Action taken: {action}')

        size_of_slice = action[0] * self.remaining_inventory
        # Convert size of slice into whole number
        size_of_slice = int(np.ceil(size_of_slice))

        # Scale action[1] to a desired range, e.g., 1 to 10
        max_time_step = 10  # Adjust based on your requirements
        timing_of_slice = int(np.ceil(action[1]))
        print(f'timing_of_slice: {timing_of_slice}')

        execution_price = self._take_action(size_of_slice)

        # Print current and next step for debugging
        print(f'Current step: {self.current_step}')
        self.current_step += timing_of_slice
        self.elapsed_time += timing_of_slice
        print(f'Next step: {self.current_step}')

        # Ensure the index is sequential
        if self.current_step >= len(self.data):
            self.current_step = len(self.data) - 1

        done = self.current_step >= len(self.data) or self.remaining_inventory <= 0
        reward = self._calculate_reward(size_of_slice, execution_price, timing_of_slice)
        self.cumulative_reward += reward
        if done:
            print(f'Cumulative Rewards: {self.cumulative_reward}')

        trade_info = {
            'step': self.current_step,
            'timestamp': self.data.index[self.current_step],
            'action': action,
            'price': execution_price,
            'shares': size_of_slice,
            'reward': reward
        }
        self.trades.append(trade_info)

        info = {
            'step': self.current_step,
            'action': action,
            'price': execution_price
        }

        return self._get_state(), reward, done, info

    
    def _add_noise_to_action(self, action):
        noise = np.random.normal(0, 0.05, size=action.shape)
        action = action + noise
        action = np.clip(action, self.action_space.low, self.action_space.high)
        return action
    
    def _take_action(self, size_of_slice):
        self.remaining_inventory -= size_of_slice
        print(f'Remaining inventory: {self.remaining_inventory}')
        if self.remaining_inventory < 0:
            self.remaining_inventory = 0
        execution_price = self.data['close'].iloc[self.current_step]
        return execution_price
    
    def _calculate_transaction_cost(self, volume, daily_volume, volatility=None):
        if volatility is None:
            volatility = self.data['volatility'].iloc[self.current_step]
        return volatility * np.sqrt(volume / daily_volume)
            
    def _calculate_reward(self, size_of_slice, execution_price, timing_of_slice):
        # Constants
        kappa = 0.1

        expected_price = self.data['expected_price'].iloc[self.current_step]
        actual_price = execution_price
        order_size = size_of_slice
        market_liquidity = self.data['market_liquidity'].iloc[self.current_step]
        time_remaining = self.preferred_timeframe - self.elapsed_time
        total_time = self.preferred_timeframe

        # Calculating various components of the reward function
        slippage = expected_price - actual_price
        urgency_penalty = self.beta * ((total_time - time_remaining) / total_time) ** 2  # Quadratic scaling
        transaction_costs = self._calculate_transaction_cost(self.data.iloc[self.current_step]['volume'], self.data['volume'].mean())
        rapid_execution_penalty = self.delta * (1 / max(time_remaining, 1))

        # Print intermediate values for debugging
#         print(f"slippage: {slippage}, market_impact: {market_impact}, urgency_penalty: {urgency_penalty}, transaction_costs: {transaction_costs}, twap_penalty: {twap_penalty}, vwap_penalty: {vwap_penalty}, implementation_shortfall: {implementation_shortfall}, rapid_execution_penalty: {rapid_execution_penalty}")

        # Combining all the components to calculate the reward
        penalty = -(slippage + urgency_penalty + transaction_costs + rapid_execution_penalty)
        
        # Adding utility theory in rewards
        reward = -penalty - (2 * kappa * (penalty ** 2))
        
        print(f"Slippage: {slippage} TC: {transaction_costs} Urgency+Rapid: {urgency_penalty + rapid_execution_penalty}")

        return reward

    
    def render(self, mode='human', close=False):
        print('--------------------------------------------------')
        print(f'Steps: {self.current_step}')
        print(f'Remaining inventory: {self.remaining_inventory}')
        print(f'Cumulative reward: {self.cumulative_reward}')
        self.print_trades()

    def print_trades(self):
        trades_df = pd.DataFrame(self.trades)
        for trade in self.trades:
            print(f"Step: {trade['step']}, Timestamp: {trade['timestamp']}, Action: {trade['action']}, Price: {trade['price']}, Shares: {trade['shares']}, Reward: {trade['reward']}")
            
```

```python
# This class object.classify_scenario will return the scenario which can be passed to transformer
import numpy as np

class MetaLearner:
    def __init__(self):
        self.high_frequency_threshold = 5000
        self.low_frequency_threshold = 200

    def classify_scenario(self, transaction_size, timeframe):
        if transaction_size > self.high_frequency_threshold and timeframe <= 1440:
            return 'high_frequency'
        elif transaction_size < self.low_frequency_threshold and timeframe > 1440:
            return 'low_frequency'
        else:
            return 'medium_frequency'
```

```python
import torch
import torch.nn as nn
from stable_baselines3.common.torch_layers import BaseFeaturesExtractor
from stable_baselines3 import PPO
from stable_baselines3.common.policies import ActorCriticPolicy
from gym import spaces


class CustomTransformerModel(BaseFeaturesExtractor):
    def __init__(self, observation_space: spaces.Box, features_dim: int = 256, scenario: str = 'medium_frequency'):
        super(CustomTransformerModel, self).__init__(observation_space, features_dim)
        self.embedding = nn.Linear(observation_space.shape[0], features_dim)  # Adapt the input size if necessary
        self.scenario = scenario
        self.encoder_layer = self._get_encoder_layer(features_dim, scenario)
        self.transformer_encoder = nn.TransformerEncoder(self.encoder_layer, num_layers=6)
        self.linear = nn.Linear(features_dim, features_dim)

    def _get_encoder_layer(self, features_dim, scenario):
        if scenario == 'high_frequency':
            return nn.TransformerEncoderLayer(d_model=features_dim, nhead=32, dropout=0.05)
        elif scenario == 'low_frequency':
            return nn.TransformerEncoderLayer(d_model=features_dim, nhead=4, dropout=0.2)
        else:  # medium_frequency
            return nn.TransformerEncoderLayer(d_model=features_dim, nhead=8, dropout=0.1)

    def forward(self, observations: torch.Tensor) -> torch.Tensor:
        x = self.embedding(observations)
        x = self.transformer_encoder(x)
        x = self.linear(x)
        return x

# Define the custom policy
class CustomTransformerPolicy(ActorCriticPolicy):
    def __init__(self, observation_space, action_space, lr_schedule, scenario='medium_frequency', *args, **kwargs):
        super(CustomTransformerPolicy, self).__init__(observation_space, action_space, lr_schedule, 
                                                      features_extractor_class=CustomTransformerModel, 
                                                      features_extractor_kwargs={'features_dim': 256, 'scenario': scenario},
                                                      *args, **kwargs)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from stable_baselines3.common.torch_layers import BaseFeaturesExtractor

class UNetTransformerEncoder(nn.Module):
    def __init__(self, in_channels, out_channels, scenario='medium_frequency'):
        super(UNetTransformerEncoder, self).__init__()
        self.in_channels = in_channels
        self.out_channels = out_channels
        self.scenario = scenario

        # Define U-Net layers
        self.encoder1 = self._block(in_channels, 64)
        self.encoder2 = self._block(64, 128)
        self.encoder3 = self._block(128, 256)
        self.encoder4 = self._block(256, 512)

        self.bottleneck = self._block(512, 1024)

        self.decoder4 = self._block(1024 + 512, 512)
        self.decoder3 = self._block(512 + 256, 256)
        self.decoder2 = self._block(256 + 128, 128)
        self.decoder1 = self._block(128 + 64, out_channels)

        # Define Transformer layers
        self.encoder_layer = self._get_encoder_layer(out_channels, scenario)
        self.transformer_encoder = nn.TransformerEncoder(self.encoder_layer, num_layers=6)

    def _block(self, in_channels, out_channels):
        return nn.Sequential(
            nn.Conv1d(in_channels, out_channels, kernel_size=3, padding=1),
            nn.BatchNorm1d(out_channels),
            nn.ReLU(inplace=True),
            nn.Conv1d(out_channels, out_channels, kernel_size=3, padding=1),
            nn.BatchNorm1d(out_channels),
            nn.ReLU(inplace=True)
        )

    def _get_encoder_layer(self, features_dim, scenario):
        if scenario == 'high_frequency':
            return nn.TransformerEncoderLayer(d_model=features_dim, nhead=32, dropout=0.05)
        elif scenario == 'low_frequency':
            return nn.TransformerEncoderLayer(d_model=features_dim, nhead=4, dropout=0.2)
        else:  # medium_frequency
            return nn.TransformerEncoderLayer(d_model=features_dim, nhead=8, dropout=0.1)

    def forward(self, x):
        enc1 = self.encoder1(x)
        enc2 = self.encoder2(F.max_pool1d(enc1, 2))
        enc3 = self.encoder3(F.max_pool1d(enc2, 2))
        enc4 = self.encoder4(F.max_pool1d(enc3, 2))

        bottleneck = self.bottleneck(F.max_pool1d(enc4, 2))

        dec4 = self.decoder4(torch.cat((F.interpolate(bottleneck, scale_factor=2), enc4), dim=1))
        dec3 = self.decoder3(torch.cat((F.interpolate(dec4, scale_factor=2), enc3), dim=1))
        dec2 = self.decoder2(torch.cat((F.interpolate(dec3, scale_factor=2), enc2), dim=1))
        dec1 = self.decoder1(torch.cat((F.interpolate(dec2, scale_factor=2), enc1), dim=1))

        # Transformer encoding
        x = self.transformer_encoder(dec1.permute(2, 0, 1)).permute(1, 2, 0)  # Permute for transformer encoder and back

        return x
```

```python
class CustomUNetTransformerModel(BaseFeaturesExtractor):
    def __init__(self, observation_space: spaces.Box, features_dim: int = 256, scenario: str = 'medium_frequency'):
        super(CustomUNetTransformerModel, self).__init__(observation_space, features_dim)
        self.embedding = nn.Linear(observation_space.shape[0], features_dim)  # Adapt the input size if necessary
        self.scenario = scenario
        self.unet_transformer = UNetTransformerEncoder(1, features_dim, scenario)

    def forward(self, observations: torch.Tensor) -> torch.Tensor:
        x = self.embedding(observations)
        x = x.unsqueeze(1)
        x = self.unet_transformer(x)
        x = x.mean(dim=2)
        return x
```

```python
# This class object.classify_scenario will return the scenario which can be passed to transformer
import numpy as np

class MetaLearner:
    def __init__(self):
        self.high_frequency_threshold = 5000
        self.low_frequency_threshold = 200

    def classify_scenario(self, transaction_size, timeframe):
        if transaction_size >= self.high_frequency_threshold and timeframe <= 1440:
            return 'high_frequency'
        elif transaction_size < self.low_frequency_threshold and timeframe > 1440:
            return 'low_frequency'
        else:
            return 'medium_frequency'
```

```python
import gym
from gym import spaces
import numpy as np
import pandas as pd

class TradingEnvironment(gym.Env):
    metadata = {'render.modes': ['human']}
    
    # Preferred timeframe: The number of steps that the user wants to complete the trade in
    def __init__(self, data,scenario, daily_trading_limit=10000, preferred_timeframe=10000, initial_inventory=1000, max_sequence_length=300):
        super(TradingEnvironment, self).__init__()
        self.data = data
        self.daily_trading_limit = daily_trading_limit
        self.current_step = 0
        
        self.preferred_timeframe = preferred_timeframe
        self.initial_inventory = initial_inventory
        self.remaining_inventory = self.initial_inventory
        self.elapsed_time = 0
        self.trades = []
        self.cumulative_reward = 0
        self.max_sequence_length = max_sequence_length
        self.scenario = scenario
        
        # Define scenario-specific penalties
        if self.scenario == 'high_frequency':
            self.beta = 1
            self.delta = 1
        elif self.scenario == 'low_frequency':
            self.beta = 1e5
            self.delta = 1e5
        else:  # medium_frequency
            self.beta = 1e3
            self.delta = 1e3

        # Extract state columns - have to change according to requirement
        self.state_columns = ['close', 'volume', 'RSI', 'MACD', 'MACD_signal', 'MACD_hist', 'Stoch_k', 'Stoch_d',
                              'OBV', 'Upper_BB', 'Middle_BB', 'Lower_BB', 'ATR_1', 'ADX', '+DI', '-DI', 'CCI', 'transaction_cost',
                              'forecast_1_hour','forecast_2_hours','forecast_3_hours','forecast_1_day','forecast_1_week',
                              'forecast_2_weeks','forecast_1_month']
        
        # Define action space - [%_slice, timing_of_next_slice] * Sequence_Length
        self.action_space = spaces.Box(low=np.tile([0.01, 10], self.max_sequence_length), high=np.tile([0.20, 100], self.max_sequence_length), dtype=np.float32)
#         self.action_space = spaces.Box(low=0, high=100, shape=(2 * self.max_sequence_length,), dtype=np.float32)
        self.observation_space = spaces.Box(low=-np.inf, high=np.inf, shape=(len(self.state_columns) + 1,), dtype=np.float32)
        
    
    def _get_state(self):
        market_conditions = self._next_observation()
        state = np.append(market_conditions, self.remaining_inventory / self.initial_inventory)
        return state
    
    def _next_observation(self):
        return self.data[self.state_columns].iloc[self.current_step].values
    
    def reset(self):
        print('------------------------------------------------Class resetted------------------------------------------------')
        self.current_step = 0
        self.cumulative_reward = 0
        self.remaining_inventory = self.initial_inventory
        self.elapsed_time = 0
        self.trades = []
        return self._get_state()
    
    def step(self, actions):
        total_reward = 0
        print('Length of actions: ', len(actions))
        # Execute each action in the sequence
        for i in range(0, len(actions), 2):
            if self.remaining_inventory <= 0 or self.current_step >= len(self.data):
                break

            action = self._add_noise_to_action(actions[i:i+2])
            print('Action: ', action)

            size_of_slice = action[0] * self.remaining_inventory
            size_of_slice = int(np.ceil(size_of_slice))

            timing_of_slice = int(np.ceil(action[1]))
            self.elapsed_time += timing_of_slice

            if self.elapsed_time > self.preferred_timeframe:
                break

            execution_price = self._take_action(size_of_slice)

            self.current_step += timing_of_slice
            

            if self.current_step >= len(self.data):
                self.current_step = len(self.data) - 1

            reward = self._calculate_reward(size_of_slice, execution_price, timing_of_slice)
            total_reward += reward
            self.cumulative_reward += reward

            trade_info = {
                'step': self.current_step,
                'timestamp': self.data.iloc[self.current_step]['timestamp'],
                'action': action,
                'price': execution_price,
                'shares': size_of_slice,
                'reward': reward,
                'inventory': self.remaining_inventory,
                'time left': self.preferred_timeframe - self.elapsed_time
            }
            self.trades.append(trade_info)

        done = self.current_step >= len(self.data) or self.remaining_inventory <= 0 or self.elapsed_time >= self.preferred_timeframe

        # Include a penalty for any remaining inventory at the end of the timeframe
        if done and self.remaining_inventory > 0:
            print("Exceeded preferred timeframe, applying penalty!")
            penalty = self.remaining_inventory * 0.1  # Adjust the penalty as needed
            total_reward -= penalty
            self.cumulative_reward -= penalty
            print(f"Penalty for remaining inventory: {penalty}")

        return self._get_state(), total_reward, done, {}


    
    def _add_noise_to_action(self, action):
#         noise = np.random.normal(0, 0.2, size=action.shape)
#         action = action + noise
#         action = np.clip(action, [0.01,1], [0.20,100])
#         return action

        # Add noise to the first action (percentage of inventory)
        noise_action_0 = np.random.normal(0, 0.02, size=action[0].shape)  # Small noise for percentage
        action[0] += noise_action_0
        action[0] = np.clip(action[0], 0.01, 0.20)

        # Add noise to the second action (timing of next slice)
        noise_action_1 = np.random.normal(0, 5, size=action[1].shape)  # Setting SD to be 5% of the range (1-100)
        action[1] += noise_action_1
        action[1] = np.clip(action[1], 10, 100) 

        return action


    
    def _take_action(self, size_of_slice):
        self.remaining_inventory -= size_of_slice
#         print(f'Remaining inventory: {self.remaining_inventory}')
        if self.remaining_inventory < 0:
            self.remaining_inventory = 0
        execution_price = self.data['close'].iloc[self.current_step]
        return execution_price
    
    def _calculate_transaction_cost(self, volume, daily_volume, volatility=None):
        if volatility is None:
            volatility = self.data['volatility'].iloc[self.current_step]
        return volatility * np.sqrt(volume / daily_volume)
            
    def _calculate_reward(self, size_of_slice, execution_price, timing_of_slice):
        # Constants
        kappa = 0.1

        expected_price = self.data['expected_price'].iloc[self.current_step]
        actual_price = execution_price
        order_size = size_of_slice
        market_liquidity = self.data['market_liquidity'].iloc[self.current_step]
        time_remaining = self.preferred_timeframe - self.elapsed_time
        total_time = self.preferred_timeframe

        # Calculating various components of the reward function
        slippage = expected_price - actual_price
        transaction_costs = self.data['transaction_cost'].iloc[self.current_step]
        
        # Penalize actions taken early in the timeframe (encourage spreading actions)
        early_action_penalty = self.delta * (time_remaining / total_time) ** 2  # Quadratic scaling
        
        if self.scenario == 'high_frequency':
            small_timestep_penalty = 0
        else:
            small_timestep_penalty = 0 if timing_of_slice > 10 else 10
        
#         time_exceed_penalty = 0
#         if self.elapsed_time > self.preferred_timeframe:
#             time_exceed_penalty = 100

        # Print intermediate values for debugging
#         print(f"slippage: {slippage}, market_impact: {market_impact}, urgency_penalty: {urgency_penalty}, transaction_costs: {transaction_costs}, twap_penalty: {twap_penalty}, vwap_penalty: {vwap_penalty}, implementation_shortfall: {implementation_shortfall}, rapid_execution_penalty: {rapid_execution_penalty}")

        # Combining all the components to calculate the reward
        penalty = (slippage + transaction_costs + early_action_penalty + small_timestep_penalty)
        
        # Adding utility theory in rewards
        reward = -penalty - (2 * kappa * (penalty ** 2))
        
        print(f"Slippage: {slippage} TC: {transaction_costs} Rapid: {early_action_penalty} Time: {small_timestep_penalty}")

        return reward

    
    def render(self, mode='human', close=False):
        print('--------------------------------------------------')
        print(f'Steps: {self.current_step}')
        print(f'Remaining inventory: {self.remaining_inventory}')
        print(f'Cumulative reward: {self.cumulative_reward}')
        self.print_trades()

    def print_trades(self):
        trades_df = pd.DataFrame(self.trades)
        for trade in self.trades:
            print(f"Step: {trade['step']}, Timestamp: {trade['timestamp']}, Action: {trade['action']}, Price: {trade['price']}, Shares: {trade['shares']}, Reward: {trade['reward']}, Inventory: {trade['inventory']}, TimeLeft: {trade['time left']}")
            
```

```python
import torch
from stable_baselines3.common.callbacks import BaseCallback

class CustomLRScheduler(BaseCallback):
    def __init__(self, learning_rate, step_size, gamma, verbose=0):
        super(CustomLRScheduler, self).__init__(verbose)
        self.learning_rate = learning_rate
        self.step_size = step_size
        self.gamma = gamma
        self.scheduler = None

    def _on_training_start(self):
        optimizer = self.model.policy.optimizer
        self.scheduler = torch.optim.lr_scheduler.StepLR(optimizer, step_size=self.step_size, gamma=self.gamma)

    def _on_step(self):
        self.scheduler.step()
        return True
```
